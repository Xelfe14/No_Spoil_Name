```python
from google.cloud import aiplatform
import vertexai
from vertexai.language_models import TextGenerationModel

def generate_summary(book_name,chapter_num):
    vertexai.init(project="wagon-bootcamp-392411", location="us-central1")
    parameters = {
        "max_output_tokens": 768,
        "temperature": 0.9,
        "top_p": 0.8,
        "top_k": 40
    }
    model = TextGenerationModel.from_pretrained("text-bison@001")

    response = model.predict(
        f"summarize chapter {chapter_num} of the book {book_name} in 250 words",
        **parameters
    )
    return response.text
```

```python
generate_summary('Rujub, the Juggler',2)
```

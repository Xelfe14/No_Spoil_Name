```python
import yake
import spacy

def chapter_topic_creation(text):
    # Running the Yake
    language = "en"
    max_ngram_sizes = 3
    deduplication_thresholds = 0.1
    deduplication_algos = ['seqm']
    windowSizes = 1
    numOfKeywords = 100
    
    custom_kw_extractor = yake.KeywordExtractor(lan=language, n=max_ngram_size, dedupLim=deduplication_threshold, dedupFunc=deduplication_algo, windowsSize=windowSize, top=numOfKeywords, features=None)
    keywords = custom_kw_extractor.extract_keywords(text)

    keyword_list=[kw[0]for kw in keywords]
    
    # Applying the NER on the Keyword_list
    texto= ' '.join(keyword_list)
    # Load the pre-trained NER model
    nlp = spacy.load('en_core_web_md')
    
    # Process the text with spaCy's NER model
    doc = nlp(texto)
    persons_list=[]
    places_list=[]
    actions_mood_list=[]
    for ent in doc.ents:
        if ent.label_== "PERSON" and ent.text not in persons_list:
            persons_list.append(f'{ent.text}')
        elif ent.label_ in ["LOC","FAC","GPE"] and ent.text not in places_list:
            places_list.append(f'{ent.text}')
        elif ent.label_ =="EVENT" and ent.text not in actions_mood_list:
            actions_mood_list.append(f'{ent.text}')
       
    # Apply POS on the keyword_list
    for token in doc:
        if token.pos_== 'VERB' and token.text not in actions_mood_list:
            actions_mood_list.append(token.text)
        elif token.pos_=='PROPN' and token.text not in persons_list:
            persons_list.append(token.text)
        elif token.pos_== 'NOUN'and token.text not in places_list:
            places_list.append(token.text)
        elif token.pos_ in ['ADJ'] and token.text not in actions_mood_list:
            actions_mood_list.append(token.text)
            
    
    topic_dict={}
    topic_dict['persons']= persons_list
    topic_dict['places']= places_list
    topic_dict['action_and_mood']= actions_mood_list
    
    topic_df = pd.DataFrame({key: pd.Series(value) for key, value in topic_dict.items()}) 
    
    return topic_df          
        
        
        
 
```

```python

import spacy
from itertools import combinations




# def pre_process(titles):
#     """
#     Pre-processes titles by removing stopwords and lemmatizing text.
#     :param titles: list of strings, contains target titles,.
#     :return: preprocessed_title_docs, list containing pre-processed titles.
#     """

#     # Preprocess all the titles
#     title_docs = [nlp(x) for x in titles]
#     preprocessed_title_docs = []
#     lemmatized_tokens = []
#     for title_doc in title_docs:
#         for token in title_doc:
#             if not token.is_stop:
#                 lemmatized_tokens.append(token.lemma_)
#         preprocessed_title_docs.append(" ".join(lemmatized_tokens))
#         del lemmatized_tokens[
#             :
#             ]  # empty the lemmatized tokens list as the code moves onto a new title

#     return preprocessed_title_docs

def similarity_filter(topics):
    """
    Recursively check if titles pass a similarity filter.
    :param titles: list of strings, contains titles.
    If the function finds titles that fail the similarity test, the above param will be the function output.
    :return: this method upon itself unless there are no similar titles; in that case the feed that was passed
    in is returned.
    """

   
    nlp = spacy.load('en_core_web_md')
    # Remove similar titles
    all_topic_pairs = list(combinations(topics, 2))
    similar_topic = []
    for pair in all_topic_pairs:
        topic1 = nlp(pair[0])
        topic2 = nlp(pair[1])
        similarity = topic1.similarity(topic2)
        if similarity > 0.8:
            similar_topic.append(pair)

    topic_to_remove = []
    for a_topic in similar_topic:
        # Get the index of the first title in the pair
        index_for_removal = topics.index(a_topic[0])
        topic_to_remove.append(index_for_removal)

    # Get indices of similar titles and remove them
    similar_topic_counts = set(topic_to_remove)
    similar_topic = [
        x[1] for x in enumerate(topics) if x[0] in similar_topic_counts
    ]

    # Exit the recursion if there are no longer any similar titles
    if len(similar_topic_counts) == 0:
        return topics

    # Continue the recursion if there are still titles to remove
    else:
        # Remove similar titles from the next input
        for topic in similar_topic:
            idx = topics.index(topic)
            topics.pop(idx)
            
        return similarity_filter(topics)

```
